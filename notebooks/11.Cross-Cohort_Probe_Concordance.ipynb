```r
data_dir <- "/cross-cohort-fa/Cross-Cohort-Allergy-Biomarkers/data"
tt_114134 <- readRDS(file.path(data_dir, "tt_114134.rds"))
tt_189148 <- readRDS(file.path(data_dir, "tt_189148.rds"))

dim(tt_114134)
dim(tt_189148)
```

```r
shared_probes <- intersect(rownames(tt_114134), rownames(tt_189148))
length(shared_probes)
both_sig <- shared_probes[
  tt_114134[shared_probes, "adj.P.Val"] < 0.1 &
  tt_189148[shared_probes, "adj.P.Val"] < 0.1
]
length(both_sig)
```

```r
concordance_check <- data.frame(
  probe        = both_sig,
  logFC_infant = tt_114134[both_sig, "logFC"],
  p_infant     = tt_114134[both_sig, "P.Value"],
  padj_infant  = tt_114134[both_sig, "adj.P.Val"],
  logFC_adol   = tt_189148[both_sig, "logFC"],
  p_adol       = tt_189148[both_sig, "P.Value"],
  padj_adol    = tt_189148[both_sig, "adj.P.Val"]
)

concordance_check$direction_matched <- sign(concordance_check$logFC_infant) == sign(concordance_check$logFC_adol)
concordance_check <- concordance_check[order(concordance_check$p_infant), ]

print(concordance_check)
write.csv(concordance_check, file.path(data_dir, "probe_concordance_p0.1.csv"), row.names = FALSE)
```
